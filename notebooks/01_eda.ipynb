# StockPredictor EDA
This notebook inspects downloaded OHLCV data and engineered indicators.

```python
from pathlib import Path
import sys

# Jupyter starts in notebooks/, so add the project root for src imports.
project_root = Path.cwd().parent
if not (project_root / 'src').exists():
    project_root = Path.cwd()
sys.path.insert(0, str(project_root))

import matplotlib.pyplot as plt
import pandas as pd
from src.data_loader import load_ticker_data
from src.features import add_features

ticker = 'AAPL'
data = load_ticker_data(ticker, data_dir=str(project_root / 'data'))
features = add_features(data)
features.head()
```

```python
fig, axes = plt.subplots(2, 1, figsize=(12, 8), sharex=True)
data['Close'].plot(ax=axes[0], title=f'{ticker} close price')
features[['ma_5', 'ma_20', 'ma_50']].plot(ax=axes[0], title='Moving averages')
features[['rsi', 'macd', 'volatility']].plot(ax=axes[1], title='Indicators')
plt.tight_layout()
```
